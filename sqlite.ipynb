```python
import sqlite3

conn = sqlite3.connect('customer.db')

c = conn.cursor()

many_customers = [
    ('Barnava','Chakraborty','chakrabortybarnava@gmail.com'),
    ('Tarun','Chakraborty','chakrabortytarun@gmail.com'),
    ('Sampa','Chakraborty','chakrabortysampa@gmail.com'),
]

c.execute("""
SELECT name FROM sqlite_master 
WHERE type='table' AND name='customers'
""")

if not c.fetchone():
    c.execute("""CREATE TABLE customers(
        firstname text,
        lastname text,
        emailaddress text
    )""")


c.executemany("INSERT INTO customers VALUES (?,?,?)",many_customers)
c.execute("SELECT rowid, * FROM customers")
s = c.fetchall()

for rowno,fname,lname,email in s:
    print(f"{rowno}. {fname} {lname} -> {email}")

conn.commit()
conn.close()
```

```python
conn = sqlite3.connect('customer.db')

c = conn.cursor()

c.execute("SELECT rowid,* FROM customers WHERE emailaddress LIKE '%s%'")
s = c.fetchall()

for rowid,fname,lname,gmail in s:
    print(f"{rowid}. \t{fname} {lname} |   {gmail}")

conn.commit()
conn.close()
```

```python
import sqlite3
conn = sqlite3.connect('customer.db')
c = conn.cursor()
llist = ['Barnava','Tarun','Sampa']
for i in llist:
    c.execute("""
    DELETE FROM customers
    WHERE firstname = ?
    AND rowid NOT IN(
          SELECT MAX(rowid)
          FROM customers
          WHERE firstname = ?
          GROUP BY emailaddress
          )
    """,(i,i))
    conn.commit()
c.execute("SELECT rowid,* FROM customers")

s = c.fetchall()

for rowid,firstname,lastname,email in s:
    print(f"{rowid}. {firstname} {lastname} -> {email}")
conn.close()
```
